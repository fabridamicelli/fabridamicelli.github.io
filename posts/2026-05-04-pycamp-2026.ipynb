PyCamp is the kind of community events that will reload your batteries and refill the hope that another way of interacting with each other different from stressful competition and selfish goals is not only possible but very real. 
Here's a few concrete examples and takeaways from PyCamp 2026 to hopefully inspire you to search for other similar events/communities that you might be interested in and can be inspiring for you.

## PyCamp 101
PyCamp is a community event primarily for Python (but not only, everybody is welcome!) developers who enjoy learning with and from other humans and/or simply want to spend a few days in nature in a friendly environment.

Everything goes, you can code (or not) on self-proposed projects or someone else's projects, you can play board/role games, you can just chill/read/draw lying on the grass or simply stroll around and strike on spontaneous conversations with people – it's up to you.

The event has a long history, but this time around I attended the [PyCampES 2026 Edition](https://pycamp.es/) in Seville, Spain.

## Learning like a child
I find the intense, continous, playful learning that we all experience during childhood rather difficult to replicate in our daily routine.  
PyCamp is a great environment to foster and unleash that kind of "working mode" we all still carry inside. I think it is not only a lot of fun, but indeed a _highly productive_ way of getting better at the craft (in this case) of dealing with computers and solving problems.

## Ignorance is a bless
I saw people (including myself) enjoying being _ignorant_.  
Yes, you read correctly.
Ignorance and openness about it was the perfect opportunity to _ask someone_ for help, to trigger a human interaction.
Chances were that the learning went even beyond one's initial question/problem and, most importantly, we _felt thankful and happy about it_.

## Hands-on
The (non-)technical topics and projects that people work on at PyCamp are absolutely free and up to the participants to decide, plan and organize their development.

There are frontal talks, are tutorials, debates, games, coding.  
Just to give you a taste, these are some of the activities that kept me busy this year:

- [Micropython hands-on workshop](https://github.com/Nekmo/micropython-workshop) for beginners to finally get my hands dirty with some hardware
- Open debate about how the use of AI for coding affects our lives, jobs, skills, etc.
- Programming a [telegram bot to log workouts](https://github.com/PyCampES/logger-bot)
- Chatting about the setup of self-hosted apps (like [pypkg.guru](https://pypkg.guru/))
- Of course, playing card games (truco!)

But that wasn't of course all.
There were all sorts of other projects: [TUI for MIDI consoles](https://github.com/PyCampES/dj-tui), kubernetes workshops, [NLP for medicine](https://github.com/PyCampES/MedNERDS), robotic arms and RL, terminal games with bots, kaggle competitions, htmx workshop, 8-bit computer emulators – just to mention a few.


## My focus coding project
We implemented a telegram bot and that was the first time coding a bot for me.
It's a simple one that helps log and retrieve gym workouts in a database remotely via voice messages.
Here's a diagram of the architecture:

```{mermaid}
flowchart LR
    user([User])
    telegram(Telegram Bot)
    server([Server])
    extractor([Speech2Text + parsing])
    logger([Logger])
    db([Database])

    user -- voice --> telegram
    telegram -- webhook --> server
    server -- audio --> extractor
    extractor -- text --> server
    server -- text --> logger
    logger --entry--> db
```

In general, I enjoyed the API of the [`python-telegram-bot`](https://github.com/python-telegram-bot/python-telegram-bot) library and now that I know how to do it I already have a few more ideas for future bot-projects.

[Here's the code](https://github.com/PyCampES/logger-bot) in case you're curious and it will take you 5 minutes to spin up your own logger-bot using that code if you also need one.

## An upwards spiral
I am not gonna dwell here on old debates about the efficiency of competition vs. cooperation.
I just want to report my experience to close up with a short reflection.  

The collaboration going on at this event was beautiful and I left with the impression that such interactions meant an uplift for everybody.
I experience the emergence of a group dynamics that I'd describe as an _upwards spiral_, a vortex of incremental and accelerating improvements in the level and the quality of both technical and human interactions.

People were ambitious, but ambitious about _sharing their knowledge_, helping  __others__ strive, giving a hand when they found someone stuck, prepearing content and materials in advance to get everybody else on board and facilitate the sharing and learning experience, _being kind_.

I don't know about you, but for me that's what Humans at their best look like.

<div style="text-align: right; font-size: 40px; font-family: 'Inconsolata', monospace;">
  /Fin
</div>
    
<div style="font-family: 'Inconsolata', monospace;">
Any bugs, questions, comments, suggestions? Ping me on [twitter](https://www.twitter.com/fabridamicelli) or drop me an e-mail (fabridamicelli at gmail).  
Share this article on your favourite platform:
</div>
